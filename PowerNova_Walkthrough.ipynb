Step 3: Creating Smart Meter Data Simulator

This module simulates real-time smart electricity meter readings.

The simulator generates:

electricity usage
voltage
current
timestamp

using random values within realistic electricity ranges.

```python
import pandas as pd
import numpy as np
from datetime import datetime

def generate_live_data():

    current_time = datetime.now()

    usage = round(np.random.uniform(1.5, 8.5), 2)

    voltage = round(np.random.uniform(220, 240), 2)

    current = round(np.random.uniform(5, 15), 2)

    data = {
        "timestamp": [current_time],
        "usage_kwh": [usage],
        "voltage": [voltage],
        "current": [current]
    }

    return pd.DataFrame(data)
```

Step 4: Creating SQLite Database System
SQLite database was used to store live electricity readings generated by the simulator.

The database stores:

timestamp
electricity usages
voltage
current

This enables:

historical analysis
live dashboard updates
machine learning predictions

```python
import sqlite3
import pandas as pd

DB_NAME = "database/powernova.db"

def create_table():

    conn = sqlite3.connect(DB_NAME)

    cursor = conn.cursor()

    cursor.execute("""
        CREATE TABLE IF NOT EXISTS electricity_data (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            timestamp TEXT,
            usage_kwh REAL,
            voltage REAL,
            current REAL
        )
    """)

    conn.commit()
    conn.close()
```

Step 5: Inserting Data into Database
Description

Generated smart meter readings are continuously inserted into the database for real-time analytics.

```python
def insert_data(data):

    conn = sqlite3.connect(DB_NAME)

    cursor = conn.cursor()

    cursor.execute("""
        INSERT INTO electricity_data
        (timestamp, usage_kwh, voltage, current)
        VALUES (?, ?, ?, ?)
    """, (
        str(data["timestamp"][0]),
        float(data["usage_kwh"][0]),
        float(data["voltage"][0]),
        float(data["current"][0])
    ))

    conn.commit()
    conn.close()
```

Step 6: Fetching Data from Database
Description

Stored electricity readings are fetched from the SQLite database for:

dashboard visualization
machine learning predictions
analytics processing

```python
def fetch_data():

    conn = sqlite3.connect(DB_NAME)

    df = pd.read_sql_query(
        "SELECT * FROM electricity_data",
        conn
    )

    conn.close()

    return df
```
